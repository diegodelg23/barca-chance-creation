```python
from statsbombpy import sb
comps = sb.competitions()
laliga = comps[comps.competition_name == "La Liga"]
laliga[["competition_id", "season_id", "season_name", "competition_name"]]
```

```python
comps.competition_name.value_counts()   # every free competition: La Liga, Champions League, World Cup…
```

```python
m = sb.matches(competition_id=11, season_id=22)   # all 2010/11 Barça matches
m[["match_date", "home_team", "away_team", "home_score", "away_score"]]
```

```python
m["goal_diff"] = (m.home_score - m.away_score).abs()   # biggest wins: 8–0 at Almería!
m.sort_values("goal_diff", ascending=False).head(5)
```
